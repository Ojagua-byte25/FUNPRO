#LISTAS
Son tipos de datos que cumplen con tres caracteristicas
1. Finita: Se puede contar cuantos elementos tiene la lista.
2. Homogenea: Solo almacena un dato en especifico.
3. Ordenada o index. Sabemos que dato ocupa cada posición.

````python
n=10
s=0
m=s
sueldo=[]
for e in range(n):
  sueldo.append(int(input(f"Ingrese el sueldo del trabajador {e+1}: ")))
  s=sueldo[e]+s #acumulador
prom=s/n
print(f"El sueldo promedio de la empresa es: {prom}")
for e in range(n):
  sueldo=int(input(f"Ingrese el sueldo del trabajador {e+1}: "))
  if sueldo[e]>prom: m=m+1
print(f"{m} trabajadores ganan más que el promedio")
````
