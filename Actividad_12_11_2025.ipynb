```python
def contador():
  nt=[]
  cantidad=int(input("Ingrese la cantidad de nt: "))
  for i in range(cantidad):
    nota=float(input(f"dijite sus noticas:{i+1} "))
    nt.append(nota)
    print(nt)
  c=0
  for a in nt:
    if a>=3:
      c+=1
      print(f"van {c} noticas")
  print(f"{c} notas fueron mayores o iguales a 3")
  return cantidad,nt

def promedio(nt,cantidad):
  t=sum(nt)
  promedio=t/cantidad
  print(f"su promedio es: {promedio}")

a,b=contador()
promedio(b,a)
```
